```python
def game():
    import random
    secert_number = random.randint(1,100)
    max_attempt=8
    att=0
    
    while att<max_attempt:
        guess =int(input("enter the guess number between 1 to 100 : "))
    
        att+=1
        
        if guess<secert_number :
            print('Too low Try Again!!!')
            
        elif guess>secert_number :
            print('too high Try Again!!!')
            
        else:
            print(f'congratulations !  you guessed the secert number {secert_number}')
            print(f'your max attempt {att}')                
    
    else:
        print(f'Game over {secert_number}')
        
game()
```
