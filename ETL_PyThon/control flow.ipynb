```python
# if

if 5==5:
    print("this is equal")
else:
    print("this is false")
```

```python
age=2

if age>=18:
    print("eligible for vote")
else :
    print("not eligible")
```

```python
mark=90

if mark>=90:
    print("O")
elif mark>80:
    print("A")
elif mark>70:
    print("B")
elif mark>60:
    print("C")
else :
    print("Fail")
```

```python
num=21
if num >0:
    if num % 2 ==0:
        print("THis is even ")
    else:
        print("odd")
```

```python
day = "apple"

match day:
    case "apple":
        print("340")
    case 2:
        print("Tus")
    case 3:
        print("Wed")
    case _:
        print("Not vaild ")
```
