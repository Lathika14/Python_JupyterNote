## File Handling

```python
file =open("text.txt","r+")
print(file.read())
file.write("hii")
```

```python
file =open("text.txt","w")
file.write("This is file handling in python")
file.close()
```

```python
file =open("text.txt","a+")
file.write("This is file handling in python/n")
print(file.read())
file.close()
```

```python
with open("text.txt","a+") as f:
    f.write("\nHii")
    print(f.read())

print(f.closed) 
```

```python
file =open("text.txt","r")
print(file.readlines())
print(f.tell())
```

```python
with open("text.txt","r") as f:
    print(f.tell())
    print(f.read(3))
    print(f.tell())
    f.seek(0)
    print(f.read(5))
    print(f.tell())
    f.seek(10,0)
    print(f.read())
```

```python
with open("text.txt","a+",encoding='utf-8') as f:
    f.write("😊 இனிய Hii")
    f.seek(0)
    print(f.read())
```

```python
f=open("text.txt","r",encoding='utf-8')
print(f.read())
```
