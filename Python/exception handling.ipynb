```python
def g():
    try:
        raise ValueError('oops')
    finally:
        return 'overridden'    # swallows the ValueError entirely!
print(g())   
```

```python
import logging
logging.basicConfig(level=logging.ERROR)
logger = logging.getLogger(__name__)
 
try:
    result = 10 / 0
except ZeroDivisionError:
    logger.exception('Division failed')   # logs message + full traceback automatically
    logger.error('Division failed', exc_info=True)
```
