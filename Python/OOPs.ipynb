```python
class Animal:

    def __init__(self,name):
        self.name = name

    def sp(self):
        print(self.name," , animal class mathod")


class dog(Animal):
 
    def __init__(self,name,color):
        super().__init__(name)
        self.color=color

c1=dog('pupy','green')

print(c1.name)
print(c1.color)
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
 
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)   # calls Animal's __init__
        self.breed = breed
 
d = Dog('Rex', 'Labrador')
print(d.name, d.breed)   
```
