```python
with open("text.txt","r") as f:
    file=f.read()

file
```

```python
with open("text.txt","r") as f:
    file=f.tell()
    f.read(3)
    print(f.tell())

file
```

```python
with open("text.txt","r") as f:
    file=f.seek(11,0)
    print(f.read())
   

file
```
