## Lists
Lists store multiple values and are mutable (modifiable).

```python
names=["lathika","well","hello"]
print(names)
print(names[1])
names.append("hii")
print(names)
names.remove("well")
print(names)
print(names.sort())
print(len(names))

```

```python
num=[1,54,3,67,8]
num.sort()
print(num)
```

## Sets
Sets store unique values only.
* Creating Set
* cities = {"Chennai", "Mumbai", "Delhi"}
* Adding Elements
* cities.add("Bangalore")
* Removing Elements
* cities.remove("Delhi")

```python
names={"lathika","vicky","daan"}
names.add("sobi")
print(names)
names.remove("lathika")
print(names)
```

```python
customers = [101,102,103,101,104,102]
unique_customers = set(customers)
print(unique_customers)
```

```python
products = ["Laptop","Mouse","Laptop","Keyboard"]
unique_products = list(set(products))
print(unique_products)
```
