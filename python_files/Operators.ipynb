## Comparison Operators
Used to compare values.

* == Equal  
* != Not Equal
*  Greater Than >
* < Less Than
*  Greater Than or Equal >=
* <= Less Than or Equal

```python
number = 10 
value = 20

print("erual : ",number==value)
print("not equal : ", number!= value)
print("GreaterThen : ",number > value)
print("less then : ",number < value)
print("greaterthen equal : ",number >=value)
print("lessthen equal : ",number<=value)
```
