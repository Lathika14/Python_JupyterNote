## Fitering

```python
import numpy as np
```

```python
marks = np.array([50,100,44,45,87])
print(marks)
```

```python
print(marks[marks>=50])
```

```python
print(marks>=50)
```

```python
pass_marks=(marks[marks>=50])
print(pass_marks)
```

```python
even=(marks[marks% 2==0])
print(even)
```

```python
pass_marks=(marks[(marks>=50) & (marks<=70)])
print(pass_marks)
```

```python
pass_marks=(marks[(marks>=50) | (marks%2==0)])
print(pass_marks)
```

```python
## multidiemisional array

arr = np.array([
    [1,2,3],[4,5,6]])
print(arr)
```

```python
even =arr[arr%2==0]
print(even)   # ellame one D array mari varum
```

```python
even =arr[(arr%2==0) | (arr==1)]
print(even)
```

```python
even = np.where(arr%2==0,arr,0)
print(even)
```
