```python
plots= dict(marker='o',
         markersize=10,
         markerfacecolor='pink',
         markeredgecolor='cyan',
         markeredgewidth=2,
         linestyle='dashed',
         linewidth=2,)
plt.plot(x,y,color='r',**plots)
plt.plot(x,z,color='b',**plots)
```

```python
plots= dict(marker='o',
         markersize=10,
         markerfacecolor='pink',
         markeredgecolor='cyan',
         markeredgewidth=2,
         linestyle='dashed',
         linewidth=2,)
plt.plot(x,y,color='r',**plots)
plt.plot(x,z,color='b',**plots)

plt.title("sample table",
           fontsize=20,
           fontweight='bold',
           color = 'blue',
           family='Arial')
plt.xlabel("num")
plt.ylabel('amount')
```

## Scatter plot

```python
x=np.array([12,10,4,7,10,5,14])
y=np.array([2000,4000,5,100,3000,500,3500])

```

```python
plt.scatter(x,y,color='r')
```

```python
plt.scatter(x,y,color='r',alpha=0.5)
```

```python
plt.scatter(x,y,color='r',alpha=0.5,edgecolor='black')
```

```python
x1=np.array([19,18,14,13,10,15,10])
y1=np.array([2020,4200,500,1000,3960,5970,3590])
```

```python
plt.scatter(x,y,color='r',alpha=0.5,edgecolor='black')
plt.scatter(x1,y1,color='blue')
```

```python
plt.scatter(x,y,color='r',alpha=0.5,edgecolor='black',label = 'rate')
plt.scatter(x1,y1,color='blue',label='sample')
plt.title('Scatter plot')
plt.legend()
```
