```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        print(f'{self.name} makes a sound')
 
class Dog(Animal):        # Dog inherits from Animal
    def bark(self):
        print(f'{self.name} says Woof!')
 
d = Dog('Rex')
d.speak()   # inherited from Animal -> Rex makes a sound
d.bark()    
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self, name):
        print(f'{self.name} makes a sound')


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Calls Animal's __init__
        self.breed = breed


d = Dog('Rex', 'Labrador')

print(d.name, d.breed)
```
